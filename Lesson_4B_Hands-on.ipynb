# 2: Digital wallet simulator

```python
balance = 25000
balance += 8500
balance -= 4200
balance /= 3
print(balance)
print(type(balance))
```

```python
# bonus challenge

balance = 25000
deposit = int(input("enter deposit amount: "))
balance += deposit
withdrawal = int(input("enter withdrawal amount: "))
balance -= withdrawal 
balance /= 3 
print(balance)
print(type(balance))
```

# 3: inventory management

```python
stock = 150
stock -= 28
stock += 75
stock //= 12
print("Number of full cartons:", stock)

```

```python
# bonus challenge

stock = int(input("intial stock: "))
sold = int(input("items sold: "))
arrived = int(input("new items: "))
carton = int(input("items per carton: "))

stock -= sold
print("remaining stock:", stock)

stock += arrived
print("updated stock:", stock)

full = stock
full //= carton
print("full carton:", full)

loose = stock 
loose %= carton 
print("loose items:", loose)

```
