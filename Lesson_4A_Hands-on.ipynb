# 1: secure login system

```python
username_correct = True
password_correct = True
account_locked = False

print("access granted" if username_correct and password_correct and not account_locked else "access denied")
```

```python
# wrong password

username_correct = True
password_correct = False
account_locked = False

print(username_correct and password_correct and not account_locked)

# because the password is incorrect
```

```python
# locked account

username_correct = True
password_correct = True
account_locked = True

print(username_correct and password_correct and not account_locked)

# because the account is locked
```

```python
# incorrect username

username_correct = False
password_correct = True
account_locked = True

print(username_correct and password_correct and not account_locked)

# because the username is incorect
```

# 2: University scholarship eligibility

```python
score = 85
participated = True
low_income = False

print("Scholarship Approved" if score >= 80 and (participated or low_income) else "scholarship not approve")

```

```python
# high score but no extarcurricular participation and low-income

score = 85
participated = False
low_income = False

print(score >= 80 and (participated or low_income))

# because the student has a high score but didn't participate and is not from a low-income family
```

```python
# low score with extarcurricular participation

score = 75
participated = True
low_income = False

print(score >= 80 and (participated or low_income))

# because the score is below 80 so the student does not qualify
```

```python
# high score and low-income

score = 85
participated = False
low_income = True

print(score >= 80 and (participated or low_income))

# because the student meets the score requirement and is from a low-income family
```

```python
# high score with both additional contions true

score = 85
participated = True
low_income = True

print(score >= 80 and (participated or low_income))

# because the student meets all the scholarship requirements 
```
