# Bonus challenge

```python
# using the .format{} method

first_name = "Grace"
last_name = "Johnson"
age = "34 Years"
blood_group = "O+"
department = "Cardiology"
full_name = (first_name + " " + last_name) .upper()


print("============ CAREPLUS HOSPITAL============")

print("Patient Name:{}".format (full_name))
print("Age:{}" .format (age))
print("Blood Goup:{}" .format (blood_group))
print("Department:{}" .format (department))

print("===========================================")
```

```python
# using the f-string

first_name = "Grace"
last_name = "Johnson"
age = "34 Years"
blood_group = "O+"
department = "Cardiology"
full_name = f"{first_name} {last_name}" .upper()


print("========= CAREPLUS HOSPITAL=========")

print(f"Patient Name:", (full_name))
print(f"Age:", (age))
print(f"Blood Goup:", (blood_group))
print(f"Department:", (department))

print("====================================")
```

# Task 3: movie recommendation search engine(Enterteinment and AI industry)

```python
movie_title = " The Last Kingdom Returns"

print("Movie Title:")
print(movie_title)

print("\nFirst Character:")
print(movie_title[0])

print("\nLast Character:")
print(movie_title[-1])

print("\nFirst Word:")
print(movie_title[:3])

print("\nLast Word:")
print(movie_title[18:])

print('\nPosition of "Returns":')
print(movie_title.find("Returns"))

print("\nSplit Result:")
print(movie_title.split())
```
