# EX. 2: payroll system

```python
employee_name = 'David Musa'
department = 'Finance'
basic_salary = 185000
bonus = 35000
tax_rate = 7.5
```

```python
employee_name = 'David Musa'
department = 'Finance'
basic_salary = 185000
bonus = 35000
tax_rate = 7.5
gross_salary = basic_salary + bonus
tax_amount = gross_salary * tax_rate
net_salary = gross_salary - tax_amount 
payroll = f"""=======================================
PAYROLL SUMMARY
=========================================
employee: {employee_name}
department: {department}
basic_salary: {bonus}
gross_salary: {gross_salary}
tax (7.5): {tax_amount}
net_salary: {net_salary}
========================================="""

print(payroll)
```
